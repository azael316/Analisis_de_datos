# Ciclos For

```python
integers = [1,2,3,4,5]
```

```python
for number in integers:   #Con esto tomamos cada elemento en el objeto
    print(number)        #Imprimimos cada elemento
```

```python
for number in integers:   
    print('yep')           #Podemos repetir una acción una determinada cantidad determina acción
```

```python
for jelly in integers:   #vemos que no importa que índice se use
    print(jelly) 
```

```python
ice_cream_dict = {'name': 'Alex', 'weekly intake': 5, 'favourite ice cream': ['MCC', 'Chocolate']}
```

```python
for cream in ice_cream_dict.values():   #Podemos tomar los objetos en un diccionario
    print(cream)
```

```python
for key, value in ice_cream_dict.items(): #Podemos tomar varios objetos a la vez
    print(key, '->' ,value)
```

```python
flavors = ['Vanilla', 'Chocolate', 'Cookie']
toppings = ['Hot fudge', 'Oreos', 'Marshmellows']
```

```python
for one in flavors:
    for two in toppings:            #Podemos anidar los ciclos for
        print(one, 'topped with', two)
```
