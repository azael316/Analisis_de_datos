```python
# IF STATEMENTS

# Definimos una variable de referencia
price = 75


# Vamos a categorizar el precio segun su valor

if price < 10:
    print('The price is ' + str(price) + ' dollars. This is Cheap') #Add the str() after showing how it won't work - or use comma
elif price < 40:
    print('The price is ' + str(price) + ' dollars. This is a not very expensive')
elif price < 100:
    print('The price is ' + str(price) + ' dollars. This is somewhat expensive')
else:
    print('The price is ' + price + ' dollars. This is very expensive!')
```

```python
# Supongamos que tenemos una lista de precios para una venta, y queremos categorizarla según su precio usando listas

#Creamos las lístas que categorizan
small_sale = []
medium_sale = []
large_sale = []

#Valor de la venta
sale = 200

if sale < 20:
    small_sale.append(sale)
elif sale < 50:
    medium_sale.append(sale)
else:
    large_sale.append(sale)
    
#Ahora podemos imprimir el precio en la lista adecuada:
print('Small Sale: ', small_sale)
print('Medium Sale: ', medium_sale)
print('Large Sale: ', large_sale)
```

```python
# FOR LOOPS
sale_amount = [1,20.70,35.2,50,100,125,70.59,106.21,700]

small_sale = []
medium_sale = []
large_sale = []


for sale in sale_amount:
    if sale < 50:
        small_sale.append(sale)
    elif sale < 100:
        medium_sale.append(sale)
    else:
        large_sale.append(sale)
    

#Now we can print these out and see what we have - each sale is put into the correct lists
print('Small Sale: ', small_sale)
print('Medium Sale: ', medium_sale)
print('Large Sale: ', large_sale)
```

```python
def first_func():      # En este caso la función no tiene variables, simplemente hará una acción cuando se mande a llamar
    print('We called our first function!') # En esta parte escribimos lo que debe realizará la función

# Llamamos la funcion sin variable
first_func()
```

```python
# Ahora vamos a crear una función con variables, por ejemplo una cadena de caracteres

def first_func(name):
    print(name,'did it!')
    
first_func('Alex')
```

```python
# PRINT VS RETURN

#Aquí el ejemplo con la función print
def print_example():
    print(100)

#Si llamamos la función
print_example()

#Si escribimos una función que tenga al final la acción return, esto sustituye la función print

def return_example():
    return 100

# Llamamos la función 
return_example()
```

```python
#Sin embargo, si llamamos la primera nos dará un error al operarla
print_example() + 100
```

```python
#En el caso de la segunda nos da un valor
return_example() + 100
```

```python
#Podemos usar el modulo chardet para determinar cual es el encoding del archivo
import chardet

with open('laptop_prices.csv', 'rb') as f:
    result = chardet.detect(f.read())

print(result)
```
