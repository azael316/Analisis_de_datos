# If-Elif-Else Statements

```python
if 25 > 10:            #Condicion lógica a cumplir
    print('It worked')   #Acción
```

```python
if 25 < 10:
    print('It worked')
else:                        #Alternativa si no se cumple la condición principal
    print('It did not work') ##Accion
```

```python
if 25 < 10:
    print('If worked')
elif 2 == 10:               #Añadimos nuevas condiciones así como sus respectivas acciones
    print('elif worked')
elif 20 > 10:
    print('elif_2 worked')
else:
    print('It did not work')
```

```python
print('It worked') if 10 < 30 else print('It did not work') #Se puede escribir en una sola línea
```

```python
print('It worked') if 10 > 30 else print('It did not work')
```

```python
#Se pueden anidar las secuencias
if 25 < 10:
    print('If worked')
elif 2 == 2:               
    print('elif worked')
    print('Nested worked') if 10 < 12 else print('Nested did not worked')
elif 20 > 10:
    print('elif_2 worked')
else:
    print('It did not work')
```
