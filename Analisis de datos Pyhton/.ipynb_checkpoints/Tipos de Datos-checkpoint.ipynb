# Tipos de datos

```python
#Los números enteros
type(-12+100)
```

```python
#Los números float
type(12+12.23)     
```

```python
#Los números complejos
type(12+3j) #j juega el papel de i
```

```python
#Las booleanas o lógicas
type(True), type(False)
```

```python
#Las cadenas
x = 'Single quote'  #Esto imprime una línea de texto

y = "Double quote" #Esto también imprime en una línea

print(x)
print(y)
```

```python
#Se pueden guardar cadenas con la información del salto de línea
multiline = """
The ice cream vaquished
my longing for sweets.
"""

multiline
```

```python
type(x),type(multiline)
```

```python
#Las cadenas como objetos guardan cierttas propiedades como posiciones
print(x[0]) #Nótese que cuenta desde 0

x[2:5]   #Intervalos tipo [,)
```

```python
print(x+x+x)

x*3 #Podemos operar con potencias como abreviacion de sumar tres veces x
```

```python
#Listas
print([1,2,3]) #POdemos hacer listas de números

print(['Cookie Dough', 'Strawberry', 'Chocoleta']) #Listas de cadenas

ice_cream = ['Vanilla', 3, True, ['Scoops', 'Spoon']]    #Objetos varios
ice_cream
```

```python
#Las listas son editables
ice_cream.append('Salted Caramel') #Podemos añadir objetos al final de la lista

ice_cream
```

```python
#Tambien tienen propiedades posicionales
print(ice_cream[0]) #Comiienza desde 0

ice_cream[0] = 'Butter Pecan' #Podemos sustituir por posición
ice_cream
```

```python
#Podemos interactuar con las listas anidadas por ejemplo su posicion
print(ice_cream[3])

ice_cream[3][1]
```

```python
#Las tuplas
tuple_scoops = (1,2,3,2,1)

print(tuple_scoops)
type(tuple_scoops)
```

```python
#Estos objetos tienen propiedades de posición también
tuple_scoops[0]
tuple_scoops[2:4]
```

```python
#Las tuplas no son editables a diferencia de las listas
tuple_scoops.append(3)
```

```python
#Los conjuntos
daily_pints = {1,2,3}

print(daily_pints)
type(daily_pints)
```

```python
#A diferencia de las listas y las tuplas los conjuntos no guardan elementos repetidos
daily_pints_log = {2,9,4,7,2,1,7,4,3,7,5,3,0,9,2,3,7} #Lo podemos definir con elementos repetidos

print(daily_pints_log) #SI son números el conjunto los ordena crecientemente automaticamente
```

```python
wdpl = {32,43,1,4,9,2,5,87,93,94}

wdpl
```

```python
#Los conjuntos tienen las propiedades de conjuntos
print(daily_pints_log|wdpl)  #Esto une los conjuntos

print(daily_pints_log & wdpl) #Esto los interseca

print(daily_pints_log - wdpl) #Esto los resta

print(daily_pints_log ^ wdpl) #Diferencia simétrica
```

```python
#Los diccionarios
dict_cream = {'name': 'Alex', 'weekly intake': 5, 'favourite ice cream': ['MCC', 'Chocolate']} #Los diccionarios
                                                                                        #guardan clave y valor
print(dict_cream)

type(dict_cream)
```

```python
#Tiene las siguientes características
print(dict_cream.values()) #Podemos ver que valores tiene

print(dict_cream.keys()) #Podemos ver las claves

print(dict_cream.items()) #Podemos ver las parejas
```

```python
#A diferencia de las listas o las tuplas no guarda posiciones
dict_cream[0]
```

```python
dict_cream['name']
```

```python
#Este también es un objeto editable
dict_cream['name'] = 'Christian'

dict_cream
```

```python
#Podemos editarlo
dict_cream.update({'weight': 300}) #Añadiendo clave/valor

dict_cream
```

```python
dict_cream.update({'weight': 100}) #Editando valores existentes

dict_cream
```

```python
del dict_cream['weight'] #Borrando cosas

dict_cream
```
