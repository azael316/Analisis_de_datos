# Operadores de comparación

```python
#Operador de identidad
10 == 10 ## El resultado es un operador booleano
```

```python
#Operador de diferencia
10 != 10
```

```python
'Vanilla' == 10 #Pueden usarse con cadenas, listas, etc.
```

```python
#Operadores de orden
10 < 10
```

```python
10 <= 10
```

```python
20 > 19
```

```python
'Vanilla' < 'Chocolate'   #Se pueden comparar la cantidad de elementos de una cadena
```

# Operadores lógicos

```python
#Conjunción
(10 > 50) and (50 > 10)
```

```python
#Disyunción
(10 > 50) or (50 > 10)
```

```python
#Negación
not (50 > 10)
```

# Operadores de pertenencia

```python
print('love' in 'I love chocolate ice cream') #POdemos ananlizar cadenas

ice_cream = 'I love chocolate ice cream'  #Variables que almacenan cadenas

print('love' in ice_cream)
```

```python
scoops = [1,2,3,4,5]

6 not in scoops #Variables de listas, tuplas, conjuntos
```
