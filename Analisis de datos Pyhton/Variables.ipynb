# Variables

```python
#La forma mas sencilla de definirlas es mediante letras y un valor
x = 22

#Para verificar la definición las podemos imprimir
print(x)
```

```python
type(x) #Esta función imprime el tipo de variable almacenada
```

```python
#Podemos almacenar texto
y = 'Mint chocolate chip'

y   #Otra manera de verificar la definición de una variable es ponerla al final de todo el código
```

```python
type(y) #El tipo de variable es llamada string
```

```python
#Las variables a diferencia de otros lenguajes de programación antiguos diferencian entre mayúscula y minúscula
a = 2
A = 3
print(a)
print(A)
```

```python
#Podemos definir múltiples variables con múltiples valores al mismo tiempo
x,y,z = 'Chocolate', 'Vanilla', 'Blue'

print(x) #Se almacena en el orden respectivo
print(z)
```

```python
#También podemos almacenar múltiples variables con el mismo valor
x = y = z = 0

print(x)
print(y)
```

```python
#Podemos definir listas de objetos
ice_cream = ['Chocolate', 'Vanilla', 'Blue']

ice_cream
#Podemos definir variables con los elementos en una lista
x,y,z = ice_cream

print(x)
print(z)
```

```python
#Así como en el caso de números con sus operaciones algebraicas básicas, en cadenas podemos concatenar cadenas

x = 'Ice cream is good' + 'for stress.'

x
```

```python
x = 2 + 2  #Podemos guardar en variables los resultados de hacer alguna operación

x
```

```python
y = '2' + 2   #No se pueden sumar cadenas (aún siendo de numero) con números porque son objetos diferentes
```

```python
#Podemos imprimir variables que almacenan cadenas junto con variables que almacenan numeros
x = "I'm"
y = 2
z = 'years old.'

print(x,y,z)
```
